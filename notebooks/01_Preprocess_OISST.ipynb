# 01. Preprocess NOAA OISST

Standardize monthly time, retain the analysis record, and apply the 13 × 13 spatial smoothing.

```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

REPO_ROOT = Path("..").resolve()
SRC_DIR = REPO_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from eams_seof.config import load_config, resolve_path

CONFIG_PATH = REPO_ROOT / "configs" / "manuscript.yml"
cfg = load_config(CONFIG_PATH, repo_root=REPO_ROOT)

print("Repository:", REPO_ROOT)
print("Config:", CONFIG_PATH)
```

```python

from eams_seof.preprocessing import (
    load_oisst,
    smooth_sst,
)

raw_path = resolve_path(REPO_ROOT, cfg["paths"]["oisst_raw"])
processed_path = resolve_path(REPO_ROOT, cfg["paths"]["oisst_smoothed"])

sst = load_oisst(
    raw_path,
    start=cfg["periods"]["data_start"],
)

display(sst)
```

```python
window = int(cfg["preprocessing"]["spatial_smoothing_points"])

sst_smoothed = smooth_sst(
    sst,
    window_size=window,
)

processed_path.parent.mkdir(parents=True, exist_ok=True)
sst_smoothed.to_dataset(name="sst").to_netcdf(processed_path)

print("Saved:", processed_path)
```
