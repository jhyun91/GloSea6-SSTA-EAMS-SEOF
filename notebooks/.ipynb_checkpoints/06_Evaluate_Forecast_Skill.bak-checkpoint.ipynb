# 06. Evaluate Forecast Skill

Evaluate the 2013–2025 forecasts using SST anomalies referenced to the fixed 1993–2016 climatology.

- EAMS ACC and MSSS by target month and lead
- Seasonal ACC and MSSS over EAMS, YS, ECS, and EJS
- ACC refinement from the three-mode model relative to the mode-1 reference

For verification, **EAMS = YS ∪ ECS ∪ EJS**.


```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

REPO_ROOT = Path("..").resolve()
SRC_DIR = REPO_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from eams_seof.config import load_config, resolve_path

CONFIG_PATH = REPO_ROOT / "configs" / "manuscript.yml"
cfg = load_config(CONFIG_PATH, repo_root=REPO_ROOT)

print("Repository:", REPO_ROOT)
print("Config    :", CONFIG_PATH)
```

```python
import numpy as np
import xarray as xr

from eams_seof.verification import (
    evaluate_configs,
    load_masks,
    plot_monthly_skill,
    plot_regional_skill,
    build_acc_refinement,
    plot_acc_refinement,
)

forecast_path = resolve_path(REPO_ROOT, cfg["paths"]["forecast_output"])
mask_path = resolve_path(REPO_ROOT, cfg["paths"]["region_masks"])
skill_path = resolve_path(REPO_ROOT, cfg["paths"]["skill_output"])

mode1_forecast_path = resolve_path(
    REPO_ROOT, cfg["paths"]["mode1_forecast_output"]
)
acc_refinement_data_path = resolve_path(
    REPO_ROOT, cfg["paths"]["acc_refinement_output"]
)

eams_skill_path = resolve_path(REPO_ROOT, cfg["figs"]["eams_skill"])
regional_skill_path = resolve_path(REPO_ROOT, cfg["figs"]["regional_skill"])
acc_refinement_path = resolve_path(REPO_ROOT, cfg["figs"]["acc_refinement"])

forecast_ds = xr.open_dataset(forecast_path)
mask_ds = xr.open_dataset(mask_path)

forecast_ds
```

## Verification

$$
\mathrm{ACC}=\frac{\sum_t f_t o_t}{\sqrt{\sum_t f_t^2\sum_t o_t^2}}
$$

$$
\mathrm{MSSS}=1-\frac{\mathrm{MSE}(f,o)}{\mathrm{MSE}(0,o)}
$$

Paired-year bootstrap resampling is used for 95% confidence intervals and positive-skill support, followed by BH-FDR correction.

## 1. Forecasts and region masks

```python
observed = forecast_ds["sst_observed_fixed_base_anomaly"]

forecasts = {
    "uncorrected": forecast_ds["sst_forecast_fixed_base_anomaly"],
    "corrected": forecast_ds["sst_forecast_fixed_base_anomaly_corrected"],
}

masks = load_masks(mask_ds, observed.isel(time=0))

assert np.array_equal(
    masks["EAMS"].values,
    (masks["YS"] | masks["ECS"] | masks["EJS"]).values,
)

print("Configurations:", list(forecasts))
print("Regions       :", list(masks))
```

## 2. Compute forecast skill

```python
v = cfg["verification"]
primary_configuration = v["primary_configuration"]
sensitivity_configuration = v["sensitivity_configuration"]

skill = evaluate_configs(
    forecasts=forecasts,
    observed=observed,
    masks=masks,
    verification_period=tuple(cfg["periods"]["verification"]),
    min_samples=int(v["minimum_samples"]),
    n_boot=int(v["bootstrap_samples"]),
    min_boot_valid=int(v["minimum_bootstrap_valid"]),
    ci_level=float(v["ci_level"]),
    fdr_q=float(v["fdr_q"]),
    target_month_fdr_scope=v["target_month_fdr_scope"],
    seasonal_fdr_scope=v["seasonal_fdr_scope"],
    seed=int(v["random_seed"]),
)

skill.attrs.update({
    "forecast_source": str(forecast_path),
    "fixed_anomaly_base": str(cfg["periods"]["fixed_anomaly_base"]),
    "primary_configuration": primary_configuration,
    "eams_definition": "YS | ECS | EJS",
})

skill_path.parent.mkdir(parents=True, exist_ok=True)
skill.to_netcdf(skill_path)

print("Saved:", skill_path)
print("Primary configuration:", primary_configuration)
print("FDR q:", float(v["fdr_q"]))
```

```python
eams = skill.sel(configuration=primary_configuration, region="EAMS")
for metric in ["ACC", "MSSS"]:
    sig = eams[f"TARGET_SIG_{metric}"]
    print(f"{metric}: {int(sig.sum())} / {sig.size} significant cells")
```

## 3. EAMS skill


```python
fig_eams, axs_eams = plot_monthly_skill(
    skill,
    configuration=primary_configuration,
    region="EAMS",
    savepath=eams_skill_path,
)

print("Saved:", eams_skill_path)
```

## 4. Regional seasonal skill


```python
fig_reg, axs_reg = plot_regional_skill(
    skill,
    configuration=primary_configuration,
    regions=("EAMS", "YS", "ECS", "EJS"),
    savepath=regional_skill_path,
)

print("Saved:", regional_skill_path)
```

## 5. ACC refinement

Compare the uncorrected three-mode forecast with the mode-1 reference. The reported mean ΔACC is conditional on grid cells exceeding the configured positive-ΔACC threshold.


```python
if not mode1_forecast_path.exists():
    raise FileNotFoundError(
        f"Run notebook 05 first: {mode1_forecast_path}"
    )

refinement = build_acc_refinement(
    mode1_path=mode1_forecast_path,
    full_model_path=forecast_path,
    region_masks=masks,
    summary_path=acc_refinement_data_path,
    verification_period=tuple(cfg["periods"]["verification"]),
    threshold_quantile=float(v["acc_refinement_quantile"]),
)

fig_ref, axs_ref = plot_acc_refinement(
    refinement,
    savepath=acc_refinement_path,
)

print("Saved:", acc_refinement_path)
```

## 6. Correction sensitivity

```python
for metric in ["ACC", "MSSS", "RMSE"]:
    sensitivity = skill[f"TARGET_{metric}"].sel(
        configuration=sensitivity_configuration,
        region="EAMS",
    )
    primary = skill[f"TARGET_{metric}"].sel(
        configuration=primary_configuration,
        region="EAMS",
    )
    print(
        f"{metric:5s}  {sensitivity_configuration}={float(sensitivity.mean()): .4f}  "
        f"{primary_configuration}={float(primary.mean()): .4f}"
    )

sensitivity_bias = skill["TARGET_BIAS"].sel(
    configuration=sensitivity_configuration,
    region="EAMS",
)
primary_bias = skill["TARGET_BIAS"].sel(
    configuration=primary_configuration,
    region="EAMS",
)
print(
    f"|BIAS| {sensitivity_configuration}={float(np.abs(sensitivity_bias).mean()): .4f}  "
    f"{primary_configuration}={float(np.abs(primary_bias).mean()): .4f}"
)
```
