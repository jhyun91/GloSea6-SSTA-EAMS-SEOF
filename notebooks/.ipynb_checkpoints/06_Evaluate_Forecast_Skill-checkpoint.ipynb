# 06. Evaluate Forecast Skill

Evaluate the 2013–2025 forecasts at lead times of 1–12 months using SST anomalies referenced to the fixed 1993–2016 climatology, with additional sensitivity tests for centered ACC, anomaly persistence, and the climatological anomaly base.

- EAMS ACC and MSSS by target month and lead
- Seasonal ACC and MSSS over EAMS, YS, ECS, and EJS
- Centered-ACC sensitivity
- No-change SST-anomaly persistence benchmark
- Fixed-anomaly-base sensitivity (1993–2016 versus training-period 1993–2012)
- Local positive ACC refinement from the three-component forecast relative to the one-component reconstruction

For verification, **EAMS = YS ∪ ECS ∪ EJS**.


```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

REPO_ROOT = Path("..").resolve()
SRC_DIR = REPO_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from eams_seof.config import load_config, resolve_path

CONFIG_PATH = REPO_ROOT / "configs" / "manuscript.yml"
cfg = load_config(CONFIG_PATH, repo_root=REPO_ROOT)

print("Repository:", REPO_ROOT)
print("Config    :", CONFIG_PATH)
```

```python
import numpy as np
import xarray as xr

from eams_seof.forecast import fixed_base_anomaly
from eams_seof.verification import (
    evaluate_skill,
    evaluate_configs,
    load_masks,
    plot_monthly_skill,
    plot_regional_skill,
    build_anomaly_persistence_forecast,
    evaluate_persistence_benchmark,
    plot_persistence_benchmark,
    plot_centered_acc_sensitivity,
    plot_anomaly_base_sensitivity,
    build_acc_refinement,
    plot_acc_refinement,
)

forecast_path = resolve_path(REPO_ROOT, cfg["paths"]["forecast_output"])
mask_path = resolve_path(REPO_ROOT, cfg["paths"]["region_masks"])
skill_path = resolve_path(REPO_ROOT, cfg["paths"]["skill_output"])
sst_path = resolve_path(REPO_ROOT, cfg["paths"]["oisst_smoothed"])

mode1_forecast_path = resolve_path(
    REPO_ROOT, cfg["paths"]["mode1_forecast_output"]
)
acc_refinement_data_path = resolve_path(
    REPO_ROOT, cfg["paths"]["acc_refinement_output"]
)

eams_skill_path = resolve_path(REPO_ROOT, cfg["figs"]["eams_skill"])
regional_skill_path = resolve_path(REPO_ROOT, cfg["figs"]["regional_skill"])
acc_refinement_path = resolve_path(REPO_ROOT, cfg["figs"]["acc_refinement"])

# Additional verification products are kept local to notebook 06 so that
# manuscript.yml does not need to be changed while these diagnostics are tested.
persistence_skill_path = REPO_ROOT / "outputs/verification/persistence_benchmark.nc"
climatology_sensitivity_path = REPO_ROOT / "outputs/verification/climatology_base_sensitivity.nc"
centered_acc_path = REPO_ROOT / "outputs/png/centered_acc_sensitivity.png"
persistence_figure_path = REPO_ROOT / "outputs/png/persistence_benchmark.png"
climatology_sensitivity_figure_path = REPO_ROOT / "outputs/png/climatology_base_sensitivity.png"

forecast_ds = xr.open_dataset(forecast_path)
mask_ds = xr.open_dataset(mask_path)

forecast_ds
```

## Verification

The primary ACC is the uncentered anomaly correlation coefficient,

$$
\mathrm{ACC}=\frac{\sum_t f_t o_t}{\sqrt{\sum_t f_t^2\sum_t o_t^2}},
$$

and the centered sensitivity uses the Pearson correlation after removing the verification-period means of forecast and observation separately.

MSSS is defined relative to the zero-anomaly climatological reference,

$$
\mathrm{MSSS}=1-\frac{\mathrm{MSE}(f,o)}{\mathrm{MSE}(0,o)}.
$$

For the persistence benchmark,

$$
\hat{T}'_{\mathrm{pers}}(t,L)=T'(t-L),
$$

and the persistence skill score (PSS) is

$$
\mathrm{PSS}=1-\frac{\mathrm{MSE}(f,o)}{\mathrm{MSE}(p,o)}.
$$

Lead 0 is excluded from the forecast-evaluation figures. For the persistence benchmark, lead 0 is additionally uninformative because the persistence forecast is identical to the verifying anomaly at zero lead. Paired-year bootstrap resampling is used for confidence intervals and positive-skill support, followed by the FDR procedure configured in `manuscript.yml`.


## 1. Forecasts and region masks


```python
observed = forecast_ds["sst_observed_fixed_base_anomaly"]

forecasts = {
    "uncorrected": forecast_ds["sst_forecast_fixed_base_anomaly"],
    "corrected": forecast_ds["sst_forecast_fixed_base_anomaly_corrected"],
}

masks = load_masks(mask_ds, observed.isel(time=0))

assert np.array_equal(
    masks["EAMS"].values,
    (masks["YS"] | masks["ECS"] | masks["EJS"]).values,
)

print("Configurations:", list(forecasts))
print("Regions       :", list(masks))
```

## 2. Compute forecast skill


```python
v = cfg["verification"]
primary_configuration = v["primary_configuration"]
sensitivity_configuration = v["sensitivity_configuration"]

skill = evaluate_configs(
    forecasts=forecasts,
    observed=observed,
    masks=masks,
    verification_period=tuple(cfg["periods"]["verification"]),
    min_samples=int(v["minimum_samples"]),
    n_boot=int(v["bootstrap_samples"]),
    min_boot_valid=int(v["minimum_bootstrap_valid"]),
    ci_level=float(v["ci_level"]),
    fdr_q=float(v["fdr_q"]),
    target_month_fdr_scope=v["target_month_fdr_scope"],
    seasonal_fdr_scope=v["seasonal_fdr_scope"],
    seed=int(v["random_seed"]),
)

skill.attrs.update({
    "forecast_source": str(forecast_path),
    "fixed_anomaly_base": str(cfg["periods"]["fixed_anomaly_base"]),
    "primary_configuration": primary_configuration,
    "eams_definition": "YS | ECS | EJS",
})

skill_path.parent.mkdir(parents=True, exist_ok=True)
skill.to_netcdf(skill_path)

print("Saved:", skill_path)
print("Primary configuration:", primary_configuration)
print("FDR q (configured in Methods):", float(v["fdr_q"]))
```

```python
eams = skill.sel(configuration=primary_configuration, region="EAMS")
for metric in ["ACC", "CENTERED_ACC", "MSSS"]:
    sig = eams[f"TARGET_SIG_{metric}"].sel(lead=slice(1, None))
    print(f"{metric}: {int(sig.sum())} / {sig.size} significant cells")
```

## 3. EAMS skill


```python
fig_eams, axs_eams = plot_monthly_skill(
    skill,
    configuration=primary_configuration,
    region="EAMS",
    savepath=eams_skill_path,
    exclude_lead0=True,
)

print("Saved:", eams_skill_path)
```

## 4. Regional seasonal skill


```python
fig_reg, axs_reg = plot_regional_skill(
    skill,
    configuration=primary_configuration,
    regions=("EAMS", "YS", "ECS", "EJS"),
    savepath=regional_skill_path,
    exclude_lead0=True,
)

print("Saved:", regional_skill_path)
```

## 5. Anomaly-persistence benchmark

The benchmark is a no-change SST-anomaly forecast: the anomaly observed at initialization is retained to the target month. The primary error-based metric is PSS; positive values indicate lower MSE for the SEOF forecast than for persistence. ΔACC is retained as a complementary correspondence metric.


```python
persistence_forecast = build_anomaly_persistence_forecast(
    observed_anomaly=observed,
    forecast_template=forecasts[primary_configuration],
)

persistence_skill = evaluate_persistence_benchmark(
    model_forecast=forecasts[primary_configuration],
    persistence_forecast=persistence_forecast,
    observed=observed,
    masks=masks,
    verification_period=tuple(cfg["periods"]["verification"]),
    min_samples=int(v["minimum_samples"]),
    n_boot=int(v["bootstrap_samples"]),
    min_boot_valid=int(v["minimum_bootstrap_valid"]),
    ci_level=float(v["ci_level"]),
    fdr_q=float(v["fdr_q"]),
    target_month_fdr_scope=v["target_month_fdr_scope"],
    seasonal_fdr_scope=v["seasonal_fdr_scope"],
    seed=int(v["random_seed"]) + 400000,
)

persistence_skill.attrs.update({
    "model_configuration": primary_configuration,
    "fixed_anomaly_base": str(cfg["periods"]["fixed_anomaly_base"]),
})

persistence_skill_path.parent.mkdir(parents=True, exist_ok=True)
persistence_skill.to_netcdf(persistence_skill_path)

# Lead 0 is not a meaningful persistence benchmark.
eams_pss = persistence_skill["TARGET_PSS"].sel(region="EAMS", lead=slice(1, None))
eams_pss_sig = persistence_skill["TARGET_SIG_PSS"].sel(region="EAMS", lead=slice(1, None))
eams_dacc = persistence_skill["TARGET_DELTA_ACC"].sel(region="EAMS", lead=slice(1, None))

print("Saved:", persistence_skill_path)
print("EAMS mean PSS, leads 1–12:", float(eams_pss.mean(skipna=True)))
print("EAMS mean ΔACC, leads 1–12:", float(eams_dacc.mean(skipna=True)))
print(
    "Positive and FDR-supported PSS cells:",
    int(eams_pss_sig.sum()), "/", eams_pss_sig.size,
)
```

```python
fig_pers, axs_pers = plot_persistence_benchmark(
    persistence_skill,
    region="EAMS",
    savepath=persistence_figure_path,
    exclude_lead0=True,
)

print("Saved:", persistence_figure_path)
```

## 6. Centered-ACC sensitivity

Centered ACC removes the verification-period mean from forecast and observation separately before calculating correlation. This diagnostic tests how strongly the primary uncentered ACC depends on the fixed anomaly reference and common mean anomaly during verification.


```python
fig_centered, axs_centered = plot_centered_acc_sensitivity(
    skill,
    configuration=primary_configuration,
    region="EAMS",
    savepath=centered_acc_path,
    exclude_lead0=True,
)

uncentered = skill["TARGET_ACC"].sel(
    configuration=primary_configuration, region="EAMS"
)
centered = skill["TARGET_CENTERED_ACC"].sel(
    configuration=primary_configuration, region="EAMS"
)

for label, leads in {
    "1–3": [1, 2, 3],
    "4–6": [4, 5, 6],
    "7–12": [7, 8, 9, 10, 11, 12],
}.items():
    u = uncentered.sel(lead=leads).mean(skipna=True)
    c = centered.sel(lead=leads).mean(skipna=True)
    print(f"{label}: uncentered={float(u): .3f}, centered={float(c): .3f}, Δ={float(c-u): .3f}")

print("Saved:", centered_acc_path)
```

## 7. Fixed-anomaly-base sensitivity

Re-express the same absolute forecasts and observations relative to the **training-period 1993–2012 climatology** and repeat verification. This changes only the anomaly reference; the forecast model and causal rolling correction are not re-estimated.


```python
sst = xr.open_dataset(sst_path)["sst"].sel(
    lat=slice(*cfg["domain"]["latitude"]),
    lon=slice(*cfg["domain"]["longitude"]),
)
sea_mask = np.isfinite(sst).any("time")
training_base_period = tuple(cfg["periods"]["training"])

if primary_configuration == "corrected":
    absolute_primary = forecast_ds["sst_forecast_absolute_corrected"]
elif primary_configuration == "uncorrected":
    absolute_primary = forecast_ds["sst_forecast_absolute"]
else:
    raise ValueError(
        "primary_configuration must be 'corrected' or 'uncorrected' for anomaly-base sensitivity."
    )

training_base_forecast, training_base_observed = fixed_base_anomaly(
    absolute_forecast_target=absolute_primary,
    observed_sst=sst,
    fixed_base_period=training_base_period,
    sea_mask=sea_mask,
)

training_base_skill = evaluate_skill(
    forecast=training_base_forecast,
    observed=training_base_observed,
    masks=masks,
    verification_period=tuple(cfg["periods"]["verification"]),
    min_samples=int(v["minimum_samples"]),
    n_boot=int(v["bootstrap_samples"]),
    min_boot_valid=int(v["minimum_bootstrap_valid"]),
    ci_level=float(v["ci_level"]),
    fdr_q=float(v["fdr_q"]),
    target_month_fdr_scope=v["target_month_fdr_scope"],
    seasonal_fdr_scope=v["seasonal_fdr_scope"],
    seed=int(v["random_seed"]) + 600000,
)

reference_label = "1993–2016"
training_label = "1993–2012"
reference_skill = skill.sel(configuration=primary_configuration, drop=True)

base_sensitivity = xr.concat(
    [
        reference_skill.expand_dims(anomaly_base=[reference_label]),
        training_base_skill.expand_dims(anomaly_base=[training_label]),
    ],
    dim="anomaly_base",
    combine_attrs="drop_conflicts",
)
base_sensitivity.attrs.update({
    "model_configuration": primary_configuration,
    "reference_anomaly_base": reference_label,
    "alternate_anomaly_base": training_label,
    "interpretation": "The same absolute forecasts are re-expressed relative to two climatological bases; the model is not refit.",
})

climatology_sensitivity_path.parent.mkdir(parents=True, exist_ok=True)
base_sensitivity.to_netcdf(climatology_sensitivity_path)

for metric in ["ACC", "CENTERED_ACC", "MSSS"]:
    ref = base_sensitivity[f"TARGET_{metric}"].sel(
        anomaly_base=reference_label, region="EAMS", lead=slice(1, None)
    )
    alt = base_sensitivity[f"TARGET_{metric}"].sel(
        anomaly_base=training_label, region="EAMS", lead=slice(1, None)
    )
    print(
        f"{metric:12s} mean Δ({training_label} - {reference_label}) = "
        f"{float((alt-ref).mean(skipna=True)): .4f}"
    )

print("Saved:", climatology_sensitivity_path)
```

```python
fig_base, axs_base = plot_anomaly_base_sensitivity(
    base_sensitivity,
    reference_base=reference_label,
    alternate_base=training_label,
    region="EAMS",
    savepath=climatology_sensitivity_figure_path,
    exclude_lead0=True,
)

print("Saved:", climatology_sensitivity_figure_path)
```

## 8. Local positive ACC refinement

Compare the **uncorrected** three-component forecast with the corresponding one-component reconstruction. The one-component product retains only the first component of the aligned forecast basis; it is not an independently refitted one-mode model. Because the alignment can rotate the retained SEOF subspace, this diagnostic measures sensitivity to forecast-basis dimensionality rather than a strictly additive contribution from raw SEOF2 and SEOF3.

The reported mean ΔACC is conditional on grid cells exceeding the configured positive-ΔACC threshold and therefore identifies localized refinements rather than a net regional benefit.


```python
if not mode1_forecast_path.exists():
    raise FileNotFoundError(
        f"Run notebook 05 first: {mode1_forecast_path}"
    )

refinement = build_acc_refinement(
    mode1_path=mode1_forecast_path,
    full_model_path=forecast_path,
    region_masks=masks,
    summary_path=acc_refinement_data_path,
    verification_period=tuple(cfg["periods"]["verification"]),
    threshold_quantile=float(v["acc_refinement_quantile"]),
)

fig_ref, axs_ref = plot_acc_refinement(
    refinement,
    savepath=acc_refinement_path,
)

print("Saved:", acc_refinement_path)
```

## 9. Rolling-correction sensitivity


```python
for metric in ["ACC", "CENTERED_ACC", "MSSS", "RMSE"]:
    sensitivity = skill[f"TARGET_{metric}"].sel(
        configuration=sensitivity_configuration,
        region="EAMS",
    )
    primary = skill[f"TARGET_{metric}"].sel(
        configuration=primary_configuration,
        region="EAMS",
    )
    print(
        f"{metric:12s}  {sensitivity_configuration}={float(sensitivity.mean()): .4f}  "
        f"{primary_configuration}={float(primary.mean()): .4f}"
    )

sensitivity_bias = skill["TARGET_BIAS"].sel(
    configuration=sensitivity_configuration,
    region="EAMS",
)
primary_bias = skill["TARGET_BIAS"].sel(
    configuration=primary_configuration,
    region="EAMS",
)
print(
    f"|BIAS|       {sensitivity_configuration}={float(np.abs(sensitivity_bias).mean()): .4f}  "
    f"{primary_configuration}={float(np.abs(primary_bias).mean()): .4f}"
)
```
